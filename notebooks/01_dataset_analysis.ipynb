```python
import numpy as np
import matplotlib.pyplot as plt
import os

print("Dataset Analysis Started")
```

```python
gt_path = "../data/train/GT/000000.npy"
noisy_path = "../data/train/NoisyLR/000000.npy"

gt = np.load(gt_path)
noisy = np.load(noisy_path)

print("GT Shape:", gt.shape)
print("NoisyLR Shape:", noisy.shape)

print("GT dtype:", gt.dtype)
print("NoisyLR dtype:", noisy.dtype)
```

```python
print("GT Statistics")
print("----------------")
print("Min:", gt.min())
print("Max:", gt.max())
print("Mean:", gt.mean())
print("Std:", gt.std())


print("\nNoisyLR Statistics")
print("----------------")
print("Min:", noisy.min())
print("Max:", noisy.max())
print("Mean:", noisy.mean())
print("Std:", noisy.std())
```

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(noisy,cmap='gray')
plt.title("NoisyLR")
plt.axis("off")


plt.subplot(1,2,2)
plt.imshow(gt,cmap='gray')
plt.title("GT")
plt.axis("off")

plt.show()
```
